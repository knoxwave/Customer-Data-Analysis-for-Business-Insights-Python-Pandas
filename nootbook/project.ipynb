<h4>👥 <span style="color: #800000;">Customer Distribution by Gender</span></h4>

```python
# Count values
gender_counts = df_customer['Gender'].value_counts()

# Create figure
plt.figure(figsize=(8,5))

# Bar chart with colors
bars = plt.bar(gender_counts.index, gender_counts.values)

# Add colors manually
colors = ['#ff69b4', '#1f77b4', '#2ca02c']  # pink, blue, green
for bar, color in zip(bars, colors):
    bar.set_color(color)

# Add value labels on top
for i, value in enumerate(gender_counts.values):
    plt.text(i, value + 2, str(value), ha='center', fontsize=10, fontweight='bold')

# Titles and labels
plt.title('Gender Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Gender')
plt.ylabel('Count')

# Style improvements
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

<h4 data-section-id="ny36qn" data-start="1243" data-end="1270"><span style="color: #800000;">🌍 Top Cities by Revenue</span></h4>

```python
# Prepare data
top_cities = df_final.groupby('City')['TransactionAmount'].sum().sort_values(ascending=False).head(10)

# Reset index for seaborn
top_cities = top_cities.reset_index()

# Set style
sns.set_style("whitegrid")

# Create figure
plt.figure(figsize=(10,6))

# Barplot
ax = sns.barplot(
    x='TransactionAmount',
    y='City',
    data=top_cities,
    palette='viridis'
)

# Add value labels
for i, v in enumerate(top_cities['TransactionAmount']):
    ax.text(v, i, f"{v:,.0f}", va='center', fontsize=9)

# Titles and labels
plt.title('Top Cities by Revenue', fontsize=16, fontweight='bold')
plt.xlabel('Revenue')
plt.ylabel('City')

# Improve layout
plt.tight_layout()

plt.show()
```

<h2 data-section-id="oske9c" data-start="244" data-end="294"><span style="color: #993300;">🔥 Step 5: Customer Segmentation (RFM Analysis)</span></h2>
<h3 data-section-id="zze8wo" data-start="296" data-end="314"><span style="color: #ff00ff;">📌 Objective</span></h3>
<p data-start="315" data-end="513">RFM (Recency, Frequency, Monetary) analysis is used to segment customers based on their purchasing behavior. This helps identify high-value customers, loyal customers, and those at risk of churning.</p>


<hr data-start="515" data-end="518" />

<h2 data-section-id="1izz8a5" data-start="520" data-end="538"><span style="color: #800080;">🧠 What is RFM?</span></h2>
<ul data-start="540" data-end="695">
 	<li data-section-id="15dvu88" data-start="540" data-end="600"><strong data-start="542" data-end="558">Recency (R):</strong> How recently a customer made a purchase</li>
 	<li data-section-id="8byllo" data-start="601" data-end="647"><strong data-start="603" data-end="621">Frequency (F):</strong> How often they purchase</li>
 	<li data-section-id="o1f4xi" data-start="648" data-end="695"><strong data-start="650" data-end="667">Monetary (M):</strong> How much money they spend</li>
</ul>

<hr data-start="697" data-end="700" />

<h2 data-section-id="1ibkibd" data-start="702" data-end="734"><span style="color: #333399;">📅 Step 1: Set Reference Date</span></h2>
&nbsp;

```python
reference_date = df_final['TransactionDate'].max()
```

<h3 data-section-id="1pqekh1" data-start="817" data-end="852">🧮 <span style="color: #003366;">Step 2: Calculate RFM Metrics</span></h3>

```python
rfm = df_final.groupby('CustomerID').agg({
    'TransactionDate': lambda x: (reference_date - x.max()).days,
    'CustomerID': 'count',
    'TransactionAmount': 'sum'
})

rfm.columns = ['Recency', 'Frequency', 'Monetary']
rfm.head()
```

<h4><span style="color: #800000;">📊 Step 3: Create RFM Scores (1–5)</span></h4>

```python
rfm['R_Score'] = pd.qcut(rfm['Recency'], 5, labels=[5,4,3,2,1])
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), 5, labels=[1,2,3,4,5])
rfm['M_Score'] = pd.qcut(rfm['Monetary'], 5, labels=[1,2,3,4,5])
```

<h4>🧩<span style="color: #800000;"> Step 4: Combine RFM Scores</span></h4>

```python
rfm['RFM_Score'] = rfm['R_Score'].astype(str) + rfm['F_Score'].astype(str) + rfm['M_Score'].astype(str)
rfm.head()
```

<h4>🏷️ <span style="color: #333300;">Step 5: Customer Segments</span></h4>

```python
def segment_customer(row):
    if row['RFM_Score'] == '555':
        return 'Champions'
    elif row['R_Score'] >= 4 and row['F_Score'] >= 4:
        return 'Loyal Customers'
    elif row['R_Score'] >= 4:
        return 'Recent Customers'
    elif row['F_Score'] >= 4:
        return 'Frequent Customers'
    elif row['M_Score'] >= 4:
        return 'Big Spenders'
    else:
        return 'At Risk'

rfm['Segment'] = rfm.apply(segment_customer, axis=1)
```
